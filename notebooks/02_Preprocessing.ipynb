```python
# ==========================================================
# NeuroBridge AI
# 02_Preprocessing.ipynb
#
# Professional Data Preprocessing
# ==========================================================

import os
import warnings
warnings.filterwarnings("ignore")

import joblib
import numpy as np
import pandas as pd

from sklearn.preprocessing import LabelEncoder

print("="*70)
print("NeuroBridge AI - Data Preprocessing")
print("="*70)

# ==========================================================
# CREATE FOLDERS
# ==========================================================

os.makedirs(r"D:\AUTISM\backend\models", exist_ok=True)
os.makedirs(r"D:\AUTISM\datasets\questionaire", exist_ok=True)

# ==========================================================
# LOAD DATASET
# ==========================================================

DATASET_PATH = r"D:\AUTISM\datasets\questionaire\autism.csv"

df = pd.read_csv(DATASET_PATH)

print("\nDataset Loaded Successfully")
print(df.shape)

# ==========================================================
# COLUMN NAMES
# ==========================================================

print("\nColumns")

print(df.columns.tolist())

# ==========================================================
# REPLACE SPECIAL VALUES
# ==========================================================

df.replace("?", np.nan, inplace=True)
df.replace("", np.nan, inplace=True)
df.replace("NA", np.nan, inplace=True)
df.replace("N/A", np.nan, inplace=True)

# ==========================================================
# REMOVE DUPLICATES
# ==========================================================

duplicates = df.duplicated().sum()

print("\nDuplicate Rows :", duplicates)

df.drop_duplicates(inplace=True)

print("Remaining :", df.shape)

# ==========================================================
# REMOVE DATA LEAKAGE
# ==========================================================

if "result" in df.columns:
    print("\nDropping result column")
    df.drop(columns=["result"], inplace=True)

# ==========================================================
# TARGET COLUMN
# ==========================================================

df["Class/ASD"] = (
    df["Class/ASD"]
    .astype(str)
    .str.upper()
    .str.strip()
)

df["Class/ASD"] = df["Class/ASD"].replace({
    "YES":1,
    "NO":0
})

# ==========================================================
# FILL NUMERIC COLUMNS
# ==========================================================

numeric_columns = df.select_dtypes(include=["int64","float64"]).columns

for col in numeric_columns:

    median = df[col].median()

    df[col].fillna(median, inplace=True)

# ==========================================================
# FILL CATEGORICAL COLUMNS
# ==========================================================

categorical_columns = df.select_dtypes(include=["object"]).columns

for col in categorical_columns:

    mode = df[col].mode()

    if len(mode) > 0:
        df[col].fillna(mode[0], inplace=True)
    else:
        df[col].fillna("Unknown", inplace=True)

# ==========================================================
# ENCODE CATEGORICAL FEATURES
# ==========================================================

encoders = {}

categorical_columns = df.select_dtypes(include=["object"]).columns

for col in categorical_columns:

    encoder = LabelEncoder()

    df[col] = encoder.fit_transform(df[col].astype(str))

    encoders[col] = encoder

# ==========================================================
# REMOVE REMAINING NaN
# ==========================================================

print("\nChecking NaN Before Final Save")

print(df.isnull().sum())

print("\nTotal NaN :", df.isnull().sum().sum())

# Drop rows if any NaN still exists

df.dropna(inplace=True)

# ==========================================================
# FINAL CHECK
# ==========================================================

print("\nFinal Dataset")

print(df.shape)

print("\nRemaining NaN")

print(df.isnull().sum())

print("\nTotal Remaining NaN")

print(df.isnull().sum().sum())

assert df.isnull().sum().sum()==0

# ==========================================================
# SAVE CLEAN DATASET
# ==========================================================

CLEAN_PATH = r"D:\AUTISM\datasets\questionaire\autism_clean.csv"

df.to_csv(CLEAN_PATH,index=False)

print("\nSaved Clean Dataset")

# ==========================================================
# SAVE LABEL ENCODERS
# ==========================================================

ENCODER_PATH = r"D:\AUTISM\backend\models\label_encoders.pkl"

joblib.dump(encoders,ENCODER_PATH)

print("Saved Label Encoders")

# ==========================================================
# FEATURES
# ==========================================================

X = df.drop("Class/ASD",axis=1)

y = df["Class/ASD"]

print("\nFeature Shape :",X.shape)

print("Target Shape :",y.shape)

# ==========================================================
# CLASS DISTRIBUTION
# ==========================================================

print("\nClass Distribution")

print(df["Class/ASD"].value_counts())

# ==========================================================
# SUMMARY
# ==========================================================

print("\n"+"="*70)

print("PREPROCESSING FINISHED SUCCESSFULLY")

print("="*70)

print("""
Files Generated

1. autism_clean.csv

2. label_encoders.pkl

Ready For

03_Model_Training.ipynb
""")
```
