```python
# ==========================================================
# NeuroBridge AI
# 01_EDA.ipynb
#
# Exploratory Data Analysis
# Autism Screening Adult Dataset
# ==========================================================

# ==============================
# 1. Import Libraries
# ==============================

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)

print("="*60)
print("NeuroBridge AI - Exploratory Data Analysis")
print("="*60)



# ==============================
# 2. Load Dataset
# ==============================

DATASET_PATH = r"D:\AUTISM\datasets\questionaire\autism.csv"

df = pd.read_csv(DATASET_PATH)

print("\nDataset Loaded Successfully.\n")



# ==============================
# 3. First Five Rows
# ==============================

print("="*60)
print("FIRST FIVE ROWS")
print("="*60)

display(df.head())



# ==============================
# 4. Last Five Rows
# ==============================

print("="*60)
print("LAST FIVE ROWS")
print("="*60)

display(df.tail())



# ==============================
# 5. Dataset Shape
# ==============================

print("="*60)
print("DATASET SHAPE")
print("="*60)

print("Rows :", df.shape[0])
print("Columns :", df.shape[1])



# ==============================
# 6. Column Names
# ==============================

print("="*60)
print("COLUMN NAMES")
print("="*60)

for i, col in enumerate(df.columns, start=1):
    print(f"{i}. {col}")



# ==============================
# 7. Dataset Information
# ==============================

print("="*60)
print("DATASET INFO")
print("="*60)

df.info()



# ==============================
# 8. Missing Values
# ==============================

print("="*60)
print("MISSING VALUES")
print("="*60)

missing = df.isnull().sum()

display(missing)

print("\nTotal Missing Values :", missing.sum())



# ==============================
# 9. Duplicate Records
# ==============================

print("="*60)
print("DUPLICATE RECORDS")
print("="*60)

duplicates = df.duplicated().sum()

print("Duplicate Rows :", duplicates)



# ==============================
# 10. Statistical Summary
# ==============================

print("="*60)
print("STATISTICAL SUMMARY")
print("="*60)

display(df.describe(include="all"))



# ==============================
# 11. Unique Values
# ==============================

print("="*60)
print("UNIQUE VALUES")
print("="*60)

for col in df.columns:
    print(f"{col} : {df[col].nunique()}")



# ==============================
# 12. Data Types
# ==============================

print("="*60)
print("DATA TYPES")
print("="*60)

display(df.dtypes)



# ==============================
# 13. Class Distribution
# ==============================

print("="*60)
print("TARGET VARIABLE")
print("="*60)

display(df["Class/ASD"].value_counts())



plt.figure(figsize=(6,5))

sns.countplot(
    x="Class/ASD",
    data=df
)

plt.title("Autism Class Distribution")
plt.xlabel("Autism")
plt.ylabel("Count")

plt.show()



# ==============================
# 14. Gender Distribution
# ==============================

plt.figure(figsize=(6,5))

sns.countplot(
    x="gender",
    data=df
)

plt.title("Gender Distribution")

plt.show()



# ==============================
# 15. Age Distribution
# ==============================

plt.figure(figsize=(8,5))

sns.histplot(
    df["age"],
    bins=20,
    kde=True
)

plt.title("Age Distribution")

plt.show()



# ==============================
# 16. Ethnicity Distribution
# ==============================

plt.figure(figsize=(12,6))

sns.countplot(
    y="ethnicity",
    data=df,
    order=df["ethnicity"].value_counts().index
)

plt.title("Ethnicity Distribution")

plt.show()



# ==============================
# 17. Country Distribution
# ==============================

plt.figure(figsize=(12,10))

sns.countplot(
    y="contry_of_res",
    data=df,
    order=df["contry_of_res"].value_counts().index
)

plt.title("Country Distribution")

plt.show()



# ==============================
# 18. Autism Family History
# ==============================

plt.figure(figsize=(6,5))

sns.countplot(
    x="austim",
    data=df
)

plt.title("Family History of Autism")

plt.show()



# ==============================
# 19. Jaundice
# ==============================

plt.figure(figsize=(6,5))

sns.countplot(
    x="jundice",
    data=df
)

plt.title("Jaundice")

plt.show()



# ==============================
# 20. Numerical Correlation
# ==============================

numeric_df = df.select_dtypes(include=["number"])

plt.figure(figsize=(14,10))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    linewidth=0.5
)

plt.title("Correlation Matrix")

plt.show()



# ==============================
# 21. AQ Score Columns
# ==============================

aq_columns = [
    "A1_Score",
    "A2_Score",
    "A3_Score",
    "A4_Score",
    "A5_Score",
    "A6_Score",
    "A7_Score",
    "A8_Score",
    "A9_Score",
    "A10_Score"
]

plt.figure(figsize=(14,6))

df[aq_columns].sum().plot(kind="bar")

plt.title("AQ Question Scores")

plt.show()



# ==============================
# 22. Boxplot of Age
# ==============================

plt.figure(figsize=(8,5))

sns.boxplot(
    x=df["age"]
)

plt.title("Age Boxplot")

plt.show()



# ==============================
# 23. Check '?' Values
# ==============================

print("="*60)
print("QUESTION MARK VALUES")
print("="*60)

for col in df.columns:

    count = (df[col] == "?").sum()

    if count > 0:
        print(col, ":", count)



# ==============================
# 24. Dataset Summary
# ==============================

print("="*60)
print("EDA SUMMARY")
print("="*60)

print(f"""
Total Records           : {df.shape[0]}
Total Columns           : {df.shape[1]}
Duplicate Rows          : {duplicates}
Missing Values          : {missing.sum()}
Target Classes          :
{df['Class/ASD'].value_counts()}

EDA Completed Successfully.
""")



# ==============================
# 25. Save Dataset
# ==============================

OUTPUT_PATH = r"D:\AUTISM\datasets\questionaire\autism_raw.csv"

df.to_csv(
    OUTPUT_PATH,
    index=False
)

print("="*60)
print("Dataset saved successfully.")
print(OUTPUT_PATH)
print("="*60)
```
